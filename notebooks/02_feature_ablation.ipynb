```python
from credit_risk_api.train import load_data, FEATURES, TARGET
from sklearn.model_selection import train_test_split

df = load_data()
X_train, X_test, y_train, y_test = train_test_split(
    df[FEATURES], df[TARGET], test_size=0.2, stratify=df[TARGET], random_state=42
)
```

```python
import lightgbm as lgb
from sklearn.metrics import roc_auc_score

def auc_without(features):
    model = lgb.LGBMClassifier(n_estimators=500, learning_rate=0.05, random_state=42, verbose=-1)
    model.fit(X_train[features], y_train)
    return roc_auc_score(y_test, model.predict_proba(X_test[features])[:, 1])

baseline = auc_without(FEATURES)
drops = {f: baseline - auc_without([x for x in FEATURES if x != f]) for f in FEATURES}
sorted(drops.items(), key=lambda kv: kv[1], reverse=True)
```
